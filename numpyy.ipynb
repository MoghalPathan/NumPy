Array Reshaping -> Reshape, Ravel, flatten

```python
# Reshape

arr = np.array([1, 2, 3, 4, 5, 6])    #1D
reshaped = arr.reshape(2,3)          # to matrix
print(reshaped)
```

```python
reshape2 = reshaped.reshape(3,2)
```

```python
reshape2
```

```python
#ravel -> convet 1D array  : it changes the array 

ravel = reshape2.ravel()
ravel
```

```python
ravel[0] = 100
print(ravel)
print(reshape2)
```

```python
# flatten -> to 1D array    : instead of changing existing array it makes copy of array and perform operations

flat = reshape2.flatten()
flat
```

```python
flat[0] = 1
```

```python
flat   
```

```python
reshape2   # as u can see here it makes copy and do not change existinf array
```
